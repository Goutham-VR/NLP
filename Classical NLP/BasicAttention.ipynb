```python
import torch
import torch.nn.functional as F


# ============================================================
# 1. TOKEN EMBEDDINGS
# ============================================================

# Sentence:
#
#     I love Python
#
# Suppose each word already has an embedding of size 2.

embeddings = torch.tensor([
    [1.0, 0.0],   # I
    [0.0, 2.0],   # love
    [2.0, 1.0]    # Python
])

print("Embeddings:")
print(embeddings)

print("Shape:", embeddings.shape)
```

```python
# 2. Choose one token
# Suppose we want to calculate attention for Python.
query = embeddings[2]

print("Query:")
print(query)

#Python → Query = [2, 1]
```

```python
#3. Compare Query with every token
# "Python wants to know which other tokens are useful to it."
scores = torch.matmul(embeddings, query)
print("Attention scores:")
print(scores)
print(scores.shape)
```

#### Let's understand what happened.

For I : [1, 0] · [2, 1]
= 1×2 + 0×1
= 2

For love : [0, 2] · [2, 1]
= 0×2 + 2×1
= 2

For Python : [2, 1] · [2, 1]
= 2×2 + 1×1
= 5

So :
scores = [2, 2, 5]
This means Python has the highest similarity with itself in this tiny example.

```python
# 4. Convert scores into attention weights
attention_weights = F.softmax(scores, dim=0)

print("Attention weights:")
print(attention_weights)

print("Sum:")
print(attention_weights.sum())
```

```python
# 5. Use the weights on the embeddings
output = torch.matmul(attention_weights, embeddings)

print("Attention output:")
print(output)
```
