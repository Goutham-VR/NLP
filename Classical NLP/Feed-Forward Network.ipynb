- ### FeedForwardNetwork
- #### 1. What is a Feed-Forward Network?
- Self-Attention: Allows tokens to gather information from other tokens.
- Feed-Forward Network (FFN): Processes the information at each token position using learned transformations.

- #### Think about the sentence:
- The cat sat on the mat.
- Self-attention helps a token gather relevant information from other tokens. The FFN then transforms the resulting representation at each position.

- #### The key distinction:

- Attention mixes information across token positions. The FFN transforms features within each token position.
- The FFN uses the same learned weights at every position.

- #### 2. Traditional Feed-Forward Network

- A traditional Transformer FFN typically contains two linear layers and an activation function.
- Its formula is: FFN(x)=W2​GELU(W1​x+b1​)+b2​

```python

# demonstration simply
import torch
import torch.nn as nn

embed_dim = 4
hidden_dim = 8

ffn = nn.Sequential(nn.Linear(embed_dim, hidden_dim),
                    nn.GELU(),
                    nn.Linear(hidden_dim, embed_dim))
    
# Batch size = 1
# Sequence length = 3
# Embedding dimension = 4

x = torch.randn(1, 3, embed_dim)

output = ffn(x)

print("Input shape:", x.shape)
print("Output shape:", output.shape)
```

```python

# demonstration using class
import torch
import torch.nn as nn

embed_dim = 4
hidden_dim = 8

class FFN(nn.Module):
    def __init__(self, embed_dim, hidden_dim):
        super().__init__()
        self.inputlayer=nn.Linear(embed_dim, hidden_dim)
        self.gelu=nn.GELU()
        self.outputlayer=nn.Linear(hidden_dim, embed_dim)
    def forward(self,x):
        output=self.inputlayer(x)
        output=self.gelu(output)
        output=self.outputlayer(output)
        return output
    
# Batch size = 1
# Sequence length = 3
# Embedding dimension = 4

x = torch.randn(1, 3, embed_dim)

model=FFN(embed_dim,hidden_dim)
output = model(x)

print("Input shape:", x.shape)
print("Output shape:", output.shape)
```
