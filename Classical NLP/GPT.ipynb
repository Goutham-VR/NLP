## Input Data

```python
with open("C:\\MyTechStack\\NLP\\Data\\training.txt", "r", encoding="utf-8") as file:
    input_data = file.read()

print(input_data)
```

## Tokenizer

```python
# Train Tokenizer
from tokenizers import ByteLevelBPETokenizer

tokenizer = ByteLevelBPETokenizer()

tokenizer.train(
    files=["C:/MyTechStack/NLP/Data/training.txt"],
    vocab_size=8000,
    min_frequency=2,
    special_tokens=[
        "<s>",
        "<pad>",
        "</s>",
        "<unk>",
        "<mask>"
    ]
)

tokenizer.save_model(
    r"C:/MyTechStack/NLP/Data/tokenizer"
)
```

```python
from tokenizers import ByteLevelBPETokenizer
import os

vocab_path = r"C:/MyTechStack/NLP/Data/tokenizer/vocab.json"
merges_path = r"C:/MyTechStack/NLP/Data/tokenizer/merges.txt"

tokenizer=ByteLevelBPETokenizer(
    vocab_path,
    merges_path
)

tokens=tokenizer.encode(input_data).tokens
token_ids=tokenizer.encode(input_data).ids

print("Tokens:",tokens)
print("Token IDS:",token_ids)
```

- Here we dont need word to id and id to word becouse we are using BPE tokenizer
- token_ids = tokenizer.encode(inputdata).ids (word to id)
- generated_text = tokenizer.decode(token_ids) (id to word)
