- #### SwiGLU - Gated Feed Forward Network

```python
import torch
import torch.nn as nn
import torch.nn.functional as F

# this is a part of transformer model
class SwiGLU(nn.Module):
    def __init__(self, embed_dim, hidden_dim):
        super().__init__()
        # Gate branch
        self.gate_proj = nn.Linear(embed_dim,hidden_dim,bias=False)
        # Up branch
        self.up_proj = nn.Linear(embed_dim,hidden_dim,bias=False)
        # Project back to embedding dimension
        self.down_proj = nn.Linear(hidden_dim,embed_dim,bias=False)

    def forward(self, x):
        # First branch: projection followed by SiLU
        # SiLU stands for Sigmoid Linear Unit. It is also known as the Swish activation function.
        gate = F.silu(self.gate_proj(x))
        # Second branch: projection
        up = self.up_proj(x)
        # Combine both branches
        combined = gate * up
        # Return to the original embedding dimension
        output = self.down_proj(combined)
        return output


# Test
batch_size = 2
seq_len = 3
embed_dim = 256
hidden_dim = 1024

x = torch.randn(batch_size, seq_len, embed_dim) #[BTE]

model = SwiGLU(embed_dim, hidden_dim)

output = model(x)

print("Input shape:", x.shape)
print("Output shape:", output.shape)
```
