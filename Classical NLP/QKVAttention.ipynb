```python
import torch
import torch.nn as nn
import torch.nn.functional as F


# ============================================================
# Q / K / V ATTENTION FROM SCRATCH
# ============================================================


# ============================================================
# 1. INPUT EMBEDDINGS
# ============================================================

# Sentence:
#
#     I love Python
#
# 3 tokens
# embedding dimension = 4

x = torch.tensor([
    [1.0, 0.0, 1.0, 0.0],   # I
    [0.0, 2.0, 0.0, 2.0],   # love
    [2.0, 1.0, 2.0, 1.0]    # Python
])

print("Input:")
print(x) 

print("Input shape:", x.shape) # this is a matrix

# ============================================================
# 2. Q / K / V LINEAR LAYERS
# ============================================================

query_layer = nn.Linear(4, 4, bias=False)
key_layer   = nn.Linear(4, 4, bias=False)
value_layer = nn.Linear(4, 4, bias=False)
# ഓരോ token-ന്റെയും 4 embedding values ഉപയോഗിച്ച് പുതിയ 4 values സൃഷ്ടിക്കുന്നു.(4,4)

# ============================================================
# 3. CREATE Q / K / V
# ============================================================

Q = query_layer(x)
K = key_layer(x)
V = value_layer(x)

print("\nQ shape:", Q.shape)
print("K shape:", K.shape)
print("V shape:", V.shape)

# Attention(Q,K,V)=Softmax(Q × Kᵀ/sqrt(dk))
# dimension  0 -> Rows
# dimension  1 -> Col
# dimension -1 -> Last Dimention
# ============================================================
# 4. Q × Kᵀ
# ============================================================
scores = torch.matmul(Q,K.transpose(-2, -1)) 

print("\nRaw scores:")
print(scores)
print("Scores shape:", scores.shape)

# ============================================================
# 5. SCALE
# ============================================================
d_k = K.size(-1) # -1 select last dimension take 4 and also we need features not tokens
# in two dim input tensor format is (no.of tokens,features/vector)
# in three dim input tensor format is (batch,no.of tokens,features/vector)
# so every time -1 select features even if the dim change

# Q × Kᵀ/sqrt(dk)
scores = scores / torch.sqrt(torch.tensor(d_k, dtype=torch.float32))

print("\nScaled scores:")
print(scores)


# ============================================================
# 6. SOFTMAX
# ============================================================
# Softmax(Q × Kᵀ/sqrt(dk))
attention_weights = F.softmax(
    scores, dim=-1)

print("\nAttention weights:")
print(attention_weights)

print("Attention weights shape:",
      attention_weights.shape)

# ============================================================
# 7. WEIGHTED SUM OF VALUES
# ============================================================

output = torch.matmul(attention_weights,V)

print("\nAttention output:")
print(output)

print("Output shape:", output.shape)
```
