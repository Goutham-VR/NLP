## What is RoPE?
- RoPE stands for Rotary Positional Embeddings.
- Instead of simply adding a positional vector to each token embedding, RoPE rotates pairs of features in the Query and Key vectors according to their positions.
- Token : ['I','love','Python'] ==> Token ID : [0,1,2] ==> Token Embedding ==> Position ==> Position Embedding ==> Final Representation
- RoPE is used in your ModernGPT implementation. Let's understand why it is needed, how it works, and how it connects to Multi-Head Attention.

#### Remember your Multi-Head Attention code:

- Q = self.query_layer(x)
- K = self.key_layer(x)
- V = self.value_layer(x)

#### RoPE is applied to Q and K before calculating attention scores.
- Token Embeddings => Q, K, V => Apply RoPE to Q and K => Attention Scores => Causal Masking => Attention Weights × V => Attention Output
- V is not directly rotated by RoPE in the standard implementation.

#### Why rotate Q and K?
- Attention scores are calculated using: Scores=Q.Ktranspose/sqrt(dk)
- RoPE transforms Q and K using position-dependent rotations. This allows the attention calculation to incorporate positional relationships between tokens.

```python
# Tiny RoPE example this demonstrates the rotation operation that inspires RoPE.
import torch
import math

# A two-dimensional vector
x = torch.tensor([1.0, 0.0])

# Rotation angle: 90 degrees
theta = math.pi / 2

# Rotation matrix
rotation_matrix = torch.tensor([
    [math.cos(theta), -math.sin(theta)],
    [math.sin(theta),  math.cos(theta)]
])

# Rotate the vector
rotated_x = rotation_matrix @ x

print("Original vector:", x)
print("Rotated vector:", rotated_x)
```

- RoPE rotates pairs of features in Q and K along the final dimension of size 64.
- That means each head has: 64/2=32
- The rotation frequencies differ across these pairs, allowing positional information to be represented at different scales.
